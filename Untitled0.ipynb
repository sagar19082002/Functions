1. What is the difference between a function and a method in Python?
A function is a standalone block of code defined using def or lambda, and it is called by its name. A method is a function associated with a class and is called on an object using dot notation.
Example:

```python
# Function
def greet(name):
    return f"Hello, {name}!"

print(greet("Alice"))  # Output: "Hello, Alice!"

# Method (associated with a list object)
my_list = [1, 2, 3]
my_list.append(4)  # append() is a method of the list class.
```

2. Explain the concept of function arguments and parameters in Python.

Parameters are variables listed in the function definition.

Arguments are actual values passed to the function during a call.
Example:

```python
def add(a, b):  # `a` and `b` are parameters.
    return a + b

result = add(2, 3)  # `2` and `3` are arguments.
print(result)  # Output: 5
```

3.What are the different ways to define and call a function in Python?

Define: Using def for named functions or lambda for anonymous functions.

Call: Using positional, keyword, default, or variable-length arguments.
Example:

```python
# Define a named function
def multiply(x, y):
    return x * y

# Define a lambda function
square = lambda x: x ** 2

# Call functions
print(multiply(3, 4))          # Positional arguments → 12
print(square(5))               # Output: 25
print(multiply(y=2, x=5))      # Keyword arguments → 10
```

6. Explain the concept of generators in Python and how they are defined.
Generators are functions that yield values lazily using yield. They create iterators without storing all values in memory.

Example:

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1

gen = count_up_to(3)
print(next(gen))  # Output: 1
```

7. What are the advantages of using generators over regular functions?

Memory efficiency: Generate values on-the-fly.

Lazy evaluation: Compute values only when needed.

Example:

```python
# Generator for large datasets
def read_large_file(file):
    with open(file) as f:
        for line in f:
            yield line  # Processes one line at a time
```

10. What is the difference between map(), reduce(), and filter() functions in Python?

map(): Transforms elements via a function.

filter(): Selects elements that satisfy a condition.

reduce(): Aggregates elements into a single value (requires functools).

Example:

```python
from functools import reduce

numbers = [1, 2, 3, 4]

# Map: Square each element
print(list(map(lambda x: x**2, numbers)))  # [1, 4, 9, 16]

# Filter: Keep even numbers
print(list(filter(lambda x: x % 2 == 0, numbers)))  # [2, 4]

# Reduce: Sum all elements
print(reduce(lambda a, b: a + b, numbers))  # 10
```

4. Write a Python function that checks if a given number is prime or not from 1 to 200.




```python
def is_prime(n):
    if n <= 1 or n > 200:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example
print(is_prime(17))  # Output: True
```

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

```python
class FibonacciIterator:
    def __init__(self, terms):
        self.terms = terms
        self.a, self.b = 0, 1
        self.count = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.count >= self.terms:
            raise StopIteration
        result = self.a
        self.a, self.b = self.b, self.a + self.b
        self.count += 1
        return result

# Example
fib = FibonacciIterator(5)
print(list(fib))  # Output: [0, 1, 1, 2, 3]
```

6. Write a generator function in Python that yields the powers of 2 up to a given exponent

```python
def powers_of_two(max_exp):
    exp = 0
    while exp <= max_exp:
        yield 2 ** exp
        exp += 1

# Example
print(list(powers_of_two(3)))  # Output: [1, 2, 4, 8]
```

10. Create a Python program that uses `filter()` to remove all the vowels from a given string.

```python
def remove_vowels(s):
    vowels = {'a', 'e', 'i', 'o', 'u', 'A', 'E', 'I', 'O', 'U'}
    return ''.join(filter(lambda char: char not in vowels, s))

# Example
print(remove_vowels("Hello World"))  # Output: "Hll Wrld"
```

product of the price per item and the quantity. The product should be increased by 10,- € if the value of the

Write a Python program using lambda and map.

```python
orders = [
    ["1.Learning python, Mark Lutz", 4, 40.95],
    ["2.Programming Python , Mark Lutz", 5, 56.80],
    ["3.Head first python , Paul Barry", 3, 32.95],
    ["4.Enfuhrung in python3, Bernd Klein", 3, 24.99]
]

# Using `map()` and lambda to process orders
result = list(map(
    lambda order: (
        int(order[0].split('.')[0]),  # Extract order number
        (order[1] * order[2] + 10 if order[1] * order[2] < 100 else order[1] * order[2])
    ),
    orders
))

print(result)
```
